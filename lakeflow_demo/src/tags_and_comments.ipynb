```python
# set the input catalog and schema name
src_catalog_name = 'demo'
src_schema_name = 'gold'

# set the output catalog and schema name
dest_catalog_name = 'demo'
dest_schema_name = 'gold'
```

```python
# Retrieve the tables and their comments
tables_df = spark.sql(f"""
SELECT 
  table_name, 
  comment 
FROM system.information_schema.tables
where table_catalog = '{src_catalog_name}' and table_schema = '{src_schema_name}'
""")

# Set the table comments in a different environment
for row in tables_df.collect():
    table_name = row['table_name']
    comment = row['comment']
    spark.sql(f"COMMENT ON TABLE {dest_catalog_name}.{dest_schema_name}.{table_name} IS '{comment}'")
```

```python
# Retrieve the columns and their comments
columns_df = spark.sql(f"""
SELECT 
  table_name, 
  column_name, 
  comment 
FROM system.information_schema.columns
WHERE table_catalog = '{src_catalog_name}' AND table_schema = '{src_schema_name}'
""")

# Set the column comments in a different environment
for row in columns_df.collect():
    table_name = row['table_name']
    column_name = row['column_name']
    comment = row['comment']
    spark.sql(f"COMMENT ON COLUMN {dest_catalog_name}.{dest_schema_name}.{table_name}.{column_name} IS '{comment}'")
```

```python
# Retrieve the table tags
table_tags_df = spark.sql(f"""
SELECT 
  table_name, 
  tag_name, 
  tag_value
FROM system.information_schema.table_tags
WHERE catalog_name = '{src_catalog_name}' AND schema_name = '{src_schema_name}'
""")

# Set the table tags in a different environment
for row in table_tags_df.collect():
    table_name = row['table_name']
    tag_name = row['tag_name']
    tag_value = row['tag_value']
    if tag_value:
      spark.sql(f"SET TAG ON TABLE {dest_catalog_name}.{dest_schema_name}.{table_name} `{tag_name}` = `{tag_value}`")
    else:
      spark.sql(f"SET TAG ON TABLE {dest_catalog_name}.{dest_schema_name}.{table_name} `{tag_name}`")
```

```python
# Retrieve the column tags
column_tags_df = spark.sql(f"""
SELECT 
  table_name, 
  column_name, 
  tag_name, 
  tag_value
FROM system.information_schema.column_tags
WHERE catalog_name = '{src_catalog_name}' AND schema_name = '{src_schema_name}'
""")

# Set the column tags in a different environment
for row in column_tags_df.collect():
    table_name = row['table_name']
    column_name = row['column_name']
    tag_name = row['tag_name']
    tag_value = row['tag_value']
    if tag_value:
      spark.sql(f"SET TAG ON COLUMN {dest_catalog_name}.{dest_schema_name}.{table_name}.{column_name} `{tag_name}` = `{tag_value}`")
    else:
      spark.sql(f"SET TAG ON COLUMN {dest_catalog_name}.{dest_schema_name}.{table_name}.{column_name} `{tag_name}`")
```
