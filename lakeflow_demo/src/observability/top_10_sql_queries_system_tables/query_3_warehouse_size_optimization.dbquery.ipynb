```sql
-- SQL Warehouse size optimization

SELECT 
    workspace_id, 
    compute.warehouse_id, 
    COUNT(*) as frequency, 
    MAX(spilled_local_bytes) AS spilled_bytes
FROM system.query.history
WHERE start_time between DATE_SUB(CURRENT_DATE, 30) AND CURRENT_DATE and compute.type = 'WAREHOUSE'
AND spilled_local_bytes > 0
GROUP BY workspace_id, compute.warehouse_id 
HAVING spilled_bytes > 0
ORDER BY frequency DESC;

```
