```sql
-- Who are the top 10 users who access the selected table (catalog.schema.table)?
SELECT
    COUNT(*),
    user_identity.email
FROM
    system.access.audit
WHERE
    action_name IN ('commandSubmit', 'getTable')
    AND request_params.full_name_arg = :table
GROUP BY user_identity.email
ORDER BY COUNT(*) DESC
LIMIT 10
```
